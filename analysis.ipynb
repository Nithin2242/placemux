## Data Dictionary (first draft)

| Column      | Likely meaning                          | Data type |
|-------------|------------------------------------------|-----------|
| survived    | 1 = survived, 0 = did not survive        | int       |
| pclass      | Passenger ticket class (1st/2nd/3rd)     | int       |
| sex         | Passenger gender                         | object    |
| age         | Passenger age in years                   | float     |
| sibsp       | # siblings/spouses aboard                | int       |
| parch       | # parents/children aboard                | int       |
| fare        | Ticket fare paid                         | float     |
| embarked    | Port of embarkation (C/Q/S)              | object    |
| class       | Same as pclass, but labeled (First etc.) | object    |
| who         | man / woman / child                      | object    |
| adult_male  | True if adult male                       | bool      |
| deck        | Deck letter (many missing values)        | object    |
