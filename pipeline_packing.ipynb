```python
from pathlib import Path

project_root = Path(".")

print("Python files:")
for path in sorted(project_root.rglob("*.py")):
    if "env" not in path.parts:
        print("-", path)

print("\nNotebook files:")
for path in sorted(project_root.rglob("*.ipynb")):
    if "env" not in path.parts:
        print("-", path)

print("\nOutput directories:")
for name in ["outputs", "logs", "pipeline"]:
    path = project_root / name
    print(f"{name}: {'exists' if path.exists() else 'not found'}")
```
