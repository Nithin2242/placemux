## Core Metrics — Definitions (write before computing)

1. **Total passengers** — count of all rows in the cleaned dataset. Grain: overall (single number).
2. **Overall survival rate** — mean of `survived` (True=1/False=0) across all rows. Grain: overall.
3. **Survival rate by class** — mean of `survived`, grouped by `pclass`. Grain: per class (3 numbers).
4. **Survival rate by sex** — mean of `survived`, grouped by `sex`. Grain: per sex (2 numbers).
5. **Average fare** — mean of `fare` across all rows. Grain: overall.
6. **Average age** — mean of `age` across all rows (post-imputation from Day 4). Grain: overall.

All metrics computed on `data/titanic_cleaned.csv`. No additional filters applied 
unless stated. All aggregations are deterministic — same input file always 
produces the same output.

```python
import pandas as pd

df = pd.read_csv("data/titanic_cleaned.csv")
print(df.shape)
```

```python
total_passengers = len(df)
overall_survival_rate = df["survived"].mean()
survival_by_class = df.groupby("pclass")["survived"].mean()
survival_by_sex = df.groupby("sex")["survived"].mean()
avg_fare = df["fare"].mean()
avg_age = df["age"].mean()

print(f"Total passengers: {total_passengers}")
print(f"Overall survival rate: {overall_survival_rate:.1%}")
print(f"Average fare: ${avg_fare:.2f}")
print(f"Average age: {avg_age:.1f}")
```
