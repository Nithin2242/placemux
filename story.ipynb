```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/titanic_cleaned.csv")
```

# Who Survived the Titanic — and Why It Wasn't Random

**Takeaway:** Survival was determined far more by who you were — your sex and 
your ticket class — than by chance. Third-class men died at nearly 6x the rate 
of first-class women.

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="pclass", y="survived", hue="sex", errorbar=None)
plt.title("First-class women almost always survived. Third-class men rarely did.")
plt.ylabel("Survival rate")
plt.xlabel("Passenger class")
plt.tight_layout()
plt.show()
```

**What this shows:** 96.7% of first-class women survived. Only 15.9% of 
third-class men did — despite being the single largest group of passengers 
on the ship (277 people). This wasn't a small gap; it was the difference 
between near-certain survival and near-certain death.

**The nuance:** Family size mattered too — but only up to a point. Passengers 
traveling with 2-4 family members survived more often (up to 71%) than those 
traveling completely alone (33%) or in very large families (0% for families 
of 8+). Survival favored small, cohesive groups — not solo travelers, and not 
large ones either.

## Recommendation

If this pattern reflects unequal access to lifeboats by class and cabin 
location rather than pure chance, future maritime safety policy should 
mandate that lifeboat access and evacuation drills be **proportional across 
all cabin classes** — not concentrated near upper-deck, higher-fare cabins.

**Expected impact:** Closing even half the survival gap between 1st and 3rd 
class passengers in a similar disaster today could mean the difference 
between losing dozens vs. hundreds of lives, purely from more equitable 
evacuation access.
